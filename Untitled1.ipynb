````python
# =====================================================================
# [Step 0] 환경 설정 및 필수 패키지
# =====================================================================
!pip install -q transformers datasets scikit-learn accelerate google-genai -U

import os
import re
import glob
import random
import time
import json
import getpass
import torch
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import concurrent.futures

from google.colab import files
from datasets import Dataset
from sklearn.model_selection import train_test_split
from sklearn.cluster import KMeans
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report
from transformers import AutoTokenizer, AutoModelForSequenceClassification, AutoModel, Trainer, TrainingArguments
from google import genai

# 재현성을 위한 시드 고정
RANDOM_SEED = 2026
random.seed(RANDOM_SEED)
np.random.seed(RANDOM_SEED)
torch.manual_seed(RANDOM_SEED)
torch.cuda.manual_seed_all(RANDOM_SEED)

if not os.path.exists('/content/hansi-database'):
    !git clone https://github.com/chjk86/hansi-database.git /content/hansi-database

# =====================================================================
# [Step 1] 메타데이터 파싱 및 비대칭(Asymmetric) 샘플링
# =====================================================================
FRONTIER_FILE = "/content/hansi-database/2026_DH_poster/변새시_2차정리본.txt"
MUNZIP_DIR = "/content/hansi-database/2025_munzip_title_text_ver"

OUTPUT_DIR = "/content/meta_bert_imbalanced_classifier"
os.makedirs(OUTPUT_DIR, exist_ok=True)

def extract_pure_hanzi(text):
    clean_text = re.sub(r'<[^>]+>', '', text)
    return re.sub(r'[^\u4E00-\u9FFF]', '', clean_text)

def get_tag_content(block, tag_name):
    start_tag = f"<{tag_name}>"
    end_tag = f"</{tag_name}>"
    block_lower = block.lower()
    if start_tag.lower() in block_lower and end_tag.lower() in block_lower:
        start_idx = block_lower.find(start_tag.lower()) + len(start_tag.lower())
        end_idx = block_lower.find(end_tag.lower(), start_idx)
        return block[start_idx:end_idx].strip()
    return "없음"

frontier_data = []
frontier_texts = set()

print("[진행] 변새시(정답 데이터) 메타데이터 파싱을 시작합니다...")
with open(FRONTIER_FILE, 'r', encoding='utf-8') as f:
    content = f.read()

for block in content.split("</Poem>"):
    if "<Poem" not in block and "<poem" not in block.lower(): continue
    text_val = get_tag_content(block, "text")
    hanzi = extract_pure_hanzi(text_val)
    if hanzi and len(hanzi) >= 10:
        term_val = get_tag_content(block, "term")
        allusion_val = get_tag_content(block, "allusion")
        evidence_val = get_tag_content(block, "evidence")
        frontier_data.append({'text': hanzi, 'metadata': f"핵심시어: {term_val} | 전고: {allusion_val} | 판별근거: {evidence_val}", 'labels': 1})
        frontier_texts.add(hanzi)

random.shuffle(frontier_data)

# 정답 데이터 분할: Train/Val (85%), Test (15%)
TEST_RATIO = 0.15
test_f_count = int(len(frontier_data) * TEST_RATIO)
train_val_f = frontier_data[test_f_count:]
test_f = frontier_data[:test_f_count]

needed_train_nf = len(train_val_f)
needed_test_nf = len(test_f) * 20
total_needed_nf = needed_train_nf + needed_test_nf

non_frontier_pool = []
munzip_files = glob.glob(os.path.join(MUNZIP_DIR, "*.txt")) + glob.glob(os.path.join(MUNZIP_DIR, "*.xml"))
random.shuffle(munzip_files)

for file_path in munzip_files:
    if len(non_frontier_pool) >= total_needed_nf: break
    try:
        with open(file_path, 'r', encoding='utf-8') as f:
            content = f.read()
        parts = content.lower().split("<text>")
        for i in range(1, len(parts)):
            if "</text>" in parts[i]:
                start_idx = content.lower().find("<text>", content.lower().find(parts[i][:10])) + 6
                end_idx = content.lower().find("</text>", start_idx)
                hanzi = extract_pure_hanzi(content[start_idx:end_idx])
                if len(hanzi) >= 10 and hanzi not in frontier_texts:
                    non_frontier_pool.append({'text': hanzi, 'metadata': "핵심시어: 없음 | 전고: 없음 | 판별근거: 없음", 'labels': 0})
    except Exception: pass

random.shuffle(non_frontier_pool)
train_val_nf = non_frontier_pool[:needed_train_nf]
test_nf = non_frontier_pool[needed_train_nf : needed_train_nf + needed_test_nf]

train_val_df = pd.DataFrame(train_val_f + train_val_nf)
test_df = pd.DataFrame(test_f + test_nf) # 1:20 불균형 테스트셋

print(f"\n[데이터 구성 완료] 총 평가 데이터: {len(test_df)}건")

# =====================================================================
# [Step 2] 훈련 데이터 2차 분할 (Train / Validation)
# =====================================================================
train_df, val_df = train_test_split(train_val_df, test_size=0.176, random_state=RANDOM_SEED, stratify=train_val_df['labels'])
train_dataset = Dataset.from_pandas(train_df)
val_dataset = Dataset.from_pandas(val_df)
test_dataset = Dataset.from_pandas(test_df)

# =====================================================================
# [Step 3] BERT 모델 파인튜닝
# =====================================================================
print("\n========== [1/4] BERT 파인튜닝 시작 ==========")
MODEL_NAME = "ethanyt/guwenbert-base"
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

def tokenize_function(examples):
    tokenized = tokenizer(examples["text"], examples["metadata"], padding="max_length", truncation=True, max_length=256)
    if "token_type_ids" in tokenized: del tokenized["token_type_ids"]
    return tokenized

tokenized_train = train_dataset.map(tokenize_function, batched=True)
tokenized_val = val_dataset.map(tokenize_function, batched=True)
tokenized_test = test_dataset.map(tokenize_function, batched=True)

model = AutoModelForSequenceClassification.from_pretrained(MODEL_NAME, num_labels=2)
training_args = TrainingArguments(
    output_dir="/content/meta_checkpoints_imbalanced", eval_strategy="epoch", save_strategy="epoch", save_total_limit=1,
    learning_rate=2e-5, per_device_train_batch_size=16, per_device_eval_batch_size=16, num_train_epochs=5, weight_decay=0.01,
    load_best_model_at_end=True, metric_for_best_model="f1", seed=RANDOM_SEED
)

def compute_metrics(eval_pred):
    logits, labels = eval_pred
    predictions = np.argmax(logits, axis=-1)
    return {"f1": f1_score(labels, predictions, average='binary', zero_division=0)}

trainer = Trainer(model=model, args=training_args, train_dataset=tokenized_train, eval_dataset=tokenized_val, compute_metrics=compute_metrics)
trainer.train()
test_df['BERT_파인튜닝_예측'] = np.argmax(trainer.predict(tokenized_test).predictions, axis=-1)

# =====================================================================
# [Step 4] BERT 비지도 군집화
# =====================================================================
print("\n========== [2/4] BERT 비지도 군집화 시작 ==========")
texts_for_eval = test_df['text'].tolist()
true_labels = test_df['labels'].tolist()
unsup_model = AutoModel.from_pretrained(MODEL_NAME)
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
unsup_model.to(device)
unsup_model.eval()

embeddings = []
with torch.no_grad():
    for i in range(0, len(texts_for_eval), 16):
        inputs = tokenizer(texts_for_eval[i:i+16], padding="max_length", truncation=True, max_length=150, return_tensors="pt").to(device)
        if "token_type_ids" in inputs: del inputs["token_type_ids"]
        embeddings.extend(unsup_model(**inputs).last_hidden_state[:, 0, :].cpu().numpy())

kmeans = KMeans(n_clusters=2, random_state=RANDOM_SEED, n_init=10)
cluster_labels = kmeans.fit_predict(np.array(embeddings))
f1_match_1 = f1_score(true_labels, cluster_labels, zero_division=0)
f1_match_2 = f1_score(true_labels, 1 - cluster_labels, zero_division=0)
test_df['BERT_비지도군집_예측'] = cluster_labels if f1_match_1 >= f1_match_2 else 1 - cluster_labels

# =====================================================================
# [Step 5] LLM 평가 (유료 API 전용 병렬 처리 + 체크포인트 이어하기)
# =====================================================================
print("\n========== [3/4 & 4/4] LLM (Gemini) 평가 시작 ==========")
if 'client' not in locals():
    GEMINI_API_KEY = getpass.getpass("Gemini API 키를 붙여넣고 Enter를 누르십시오: ")
    client = genai.Client(api_key=GEMINI_API_KEY)

LLM_MODEL_LIST = ['gemini-3.6-flash', 'gemini-3.5-flash', 'gemini-3.5-flash-lite']

def call_llm_with_fallback(client, model_list, prompt, max_retries=3, base_delay=2.0):
    for model_id in model_list:
        retries, delay = 0, base_delay
        while retries <= max_retries:
            try:
                return client.models.generate_content(model=model_id, contents=prompt).text
            except Exception as e:
                error_msg = str(e).lower()
                if any(err in error_msg for err in ["503", "429", "unavailable", "quota"]):
                    if retries == max_retries: break
                    time.sleep(delay + random.uniform(0.1, 0.5))
                    retries += 1; delay *= 1.5
                else: break
    raise Exception("모든 Flash 대체 모델 호출에 실패했습니다.")

def parse_cot_response(response_text):
    cleaned = re.sub(r'^```(?:json)?\s*', '', response_text.strip(), flags=re.IGNORECASE)
    cleaned = re.sub(r'\s*```$', '', cleaned)
    try:
        data = json.loads(cleaned)
        return int(data.get("prediction", 0)), f"[시어] {data.get('term', '')} | [근거] {data.get('evidence', '')}"
    except: return 0, "JSON 파싱 실패"

def process_single_text(i, text, prompt_func):
    try:
        response_text = call_llm_with_fallback(client, LLM_MODEL_LIST, prompt_func(text))
        pred, reason = parse_cot_response(response_text)
        return i, pred, reason
    except Exception as e:
        return i, 0, f"API 오류: {e}"

def evaluate_llm_concurrently(df, text_col, result_pred_col, result_reason_col, prompt_func, checkpoint_filename, max_workers=10):
    texts = df[text_col].tolist()

    if result_pred_col not in df.columns:
        df[result_pred_col] = -1
        df[result_reason_col] = ""

    pending_indices = [i for i, val in enumerate(df[result_pred_col]) if val == -1]

    if not pending_indices:
        print(f"해당 작업({result_pred_col})은 이미 모든 데이터 처리가 완료되었습니다.")
        return

    print(f"총 {len(texts)}건 중, 이미 완료된 데이터를 제외한 잔여 데이터 {len(pending_indices)}건 병렬 처리 시작...")

    with concurrent.futures.ThreadPoolExecutor(max_workers=max_workers) as executor:
        future_to_index = {
            executor.submit(process_single_text, i, texts[i], prompt_func): i
            for i in pending_indices
        }

        completed_count = 0
        for future in concurrent.futures.as_completed(future_to_index):
            i, pred, reason = future.result()
            df.at[i, result_pred_col] = pred
            df.at[i, result_reason_col] = reason
            completed_count += 1

            if completed_count % 50 == 0 or completed_count == len(pending_indices):
                print(f" - 잔여 작업 진행 상황: {completed_count}/{len(pending_indices)} 완료 (체크포인트 갱신)")
                df.to_csv(f'/content/{checkpoint_filename}', index=False, encoding='utf-8-sig')

def get_fewshot_cot_prompt(text):
    return f"""당신은 한국 고전문학 및 한문학에 정통한 연구자입니다. 다음 한시 원문을 읽고, 변새시인지 판별하십시오.
[예시 1: 변새시]
원문: 談笑腥塵一掃空海中蛟鰐識威風指揮能事天應惜合散奇謀鬼不功夜報捷書靈夏外春收烽火浿江東尙餘蟣蝨逃湯鼎願借丹衡扇祝融
응답: {{"term": "捷書, 烽火", "evidence": "변방의 승전보(捷書)와 봉화(烽火)를 거두는 상황을 묘사.", "prediction": 1}}
[예시 2: 비변새시]
원문: 佳麗金陵聖祖居丹靑想像淚沾裾石頭獅子雲霞古玄武瓜洲舟楫踈一代衣冠宮殿在百年日月山河虛中天漢祚昔人語何迺于今久寂如
응답: {{"term": "金陵, 玄武", "evidence": "금릉의 옛 도읍을 바라보는 회고(懷古)시.", "prediction": 0}}
원문: {text}
반드시 동일한 JSON 형식으로만 응답하십시오."""

def get_zeroshot_cot_prompt(text):
    return f"""당신은 한국 고전문학 연구자입니다. 다음 한시가 변새시인지 판별하십시오.
원문: {text}
반드시 아래 JSON 형식으로만 응답하십시오.
{{"term": "추출한 시어", "evidence": "판별 근거", "prediction": 1}}"""

print("\n[제로샷 평가 진행 중...]")
evaluate_llm_concurrently(test_df, 'text', 'LLM_제로샷_예측', 'LLM_제로샷_근거', get_zeroshot_cot_prompt, 'checkpoint_zero.csv', max_workers=10)

print("\n[퓨샷 평가 진행 중...]")
evaluate_llm_concurrently(test_df, 'text', 'LLM_퓨샷_예측', 'LLM_퓨샷_근거', get_fewshot_cot_prompt, 'checkpoint_few.csv', max_workers=10)

# =====================================================================
# [Step 6] 결과 저장 및 시각화
# =====================================================================
csv_output = '/content/final_integrated_evaluation_imbalanced_cot.csv'
test_df.to_csv(csv_output, index=False, encoding='utf-8-sig')

fig, axes = plt.subplots(2, 2, figsize=(12, 11))
fig.suptitle('Frontier Poetry Classification (Imbalanced 1:20 Test Set)', fontsize=18, y=0.95)
CLASS_NAMES = ['Non-Frontier (0)', 'Frontier (1)']

plot_data = [
    (axes[0, 0], confusion_matrix(test_df['labels'], test_df['BERT_파인튜닝_예측']), 'BERT: Fine-tuned'),
    (axes[0, 1], confusion_matrix(test_df['labels'], test_df['BERT_비지도군집_예측']), 'BERT: Clustering'),
    (axes[1, 0], confusion_matrix(test_df['labels'], test_df['LLM_퓨샷_예측']), 'LLM: Few-shot'),
    (axes[1, 1], confusion_matrix(test_df['labels'], test_df['LLM_제로샷_예측']), 'LLM: Zero-shot')
]

for ax, cm, title in plot_data:
    sns.heatmap(cm, annot=True, fmt='d', cmap='Reds', xticklabels=CLASS_NAMES, yticklabels=CLASS_NAMES, annot_kws={"size": 16}, ax=ax)
    tn, fp, fn, tp = cm.ravel()
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0
    f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0
    ax.set_title(f"{title}\n(F1: {f1:.3f} | Prec: {precision:.3f} | Rec: {recall:.3f})", fontsize=13, pad=12)

plt.tight_layout(rect=[0.05, 0.05, 1, 0.93])
png_output = '/content/confusion_matrix_imbalanced.png'
plt.savefig(png_output, dpi=300, bbox_inches='tight', transparent=False)
plt.close()

files.download(csv_output)
files.download(png_output)
````
